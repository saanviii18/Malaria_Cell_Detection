```python
# ============================================================
# MALARIA CELL DETECTION — ENHANCED VERSION
# CNN + Transfer Learning (MobileNetV2) | Grad-CAM | Full Eval
# ============================================================

# ==============================
# IMPORTS
# ==============================
import os
import shutil
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import cv2

from PIL import Image
from pathlib import Path

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.callbacks import (
    EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
)
from tensorflow.keras.preprocessing import image as keras_image
from tensorflow.keras.optimizers import Adam

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

warnings.filterwarnings("ignore")
print("TensorFlow Version:", tf.__version__)

# ==============================
# REPRODUCIBILITY
# ==============================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ==============================
# PATH SETUP
# ==============================
# Download the dataset from Kaggle (iarunava/cell-images-for-detecting-malaria)
# and extract it so that data/cell_images/Parasitized and data/cell_images/Uninfected exist.
BASE_DIR = os.path.join("data", "cell_images")

parasitized_path = os.path.join(BASE_DIR, "Parasitized")
uninfected_path  = os.path.join(BASE_DIR, "Uninfected")

print("\nInitial Counts:")
print("Parasitized :", len(os.listdir(parasitized_path)))
print("Uninfected  :", len(os.listdir(uninfected_path)))

# ==============================
# REMOVE CORRUPTED IMAGES
# ==============================
def remove_corrupted(folder):
    removed = 0
    for file in os.listdir(folder):
        path = os.path.join(folder, file)
        try:
            img = Image.open(path)
            img.verify()
        except Exception:
            os.remove(path)
            removed += 1
            print(f"  Removed corrupt file: {path}")
    print(f"  {os.path.basename(folder)} → {removed} corrupt file(s) removed")

print("\nScanning for corrupted images...")
remove_corrupted(parasitized_path)
remove_corrupted(uninfected_path)

p_count = len(os.listdir(parasitized_path))
u_count = len(os.listdir(uninfected_path))
print("\nAfter Cleaning:")
print("Parasitized :", p_count)
print("Uninfected  :", u_count)

# ==============================
# CLASS IMBALANCE CHECK
# ==============================
print("\nClass Balance Check:")
total = p_count + u_count
print(f"  Parasitized : {p_count} ({100 * p_count / total:.1f}%)")
print(f"  Uninfected  : {u_count} ({100 * u_count / total:.1f}%)")
if abs(p_count - u_count) / total > 0.1:
    print("  [WARNING] Class imbalance detected. Consider class_weight in model.fit()")
else:
    print("  [OK] Classes are reasonably balanced")

# ==============================
# VISUALIZE SAMPLE IMAGES
# ==============================
def show_sample_images(parasitized_dir, uninfected_dir, n=5):
    fig, axes = plt.subplots(2, n, figsize=(15, 6))
    fig.suptitle("Sample Cell Images", fontsize=14, fontweight='bold')

    para_files  = random.sample(os.listdir(parasitized_dir), n)
    uninf_files = random.sample(os.listdir(uninfected_dir), n)

    for i, (pf, uf) in enumerate(zip(para_files, uninf_files)):
        # Row 0 — Parasitized
        img_p = cv2.imread(os.path.join(parasitized_dir, pf))
        img_p = cv2.cvtColor(img_p, cv2.COLOR_BGR2RGB)
        axes[0, i].imshow(img_p)
        axes[0, i].set_title("Parasitized", color='red', fontsize=9)
        axes[0, i].axis('off')

        # Row 1 — Uninfected
        img_u = cv2.imread(os.path.join(uninfected_dir, uf))
        img_u = cv2.cvtColor(img_u, cv2.COLOR_BGR2RGB)
        axes[1, i].imshow(img_u)
        axes[1, i].set_title("Uninfected", color='green', fontsize=9)
        axes[1, i].axis('off')

    plt.tight_layout()
    plt.savefig("sample_images.png", dpi=150)
    plt.show()
    print("Saved: sample_images.png")

show_sample_images(parasitized_path, uninfected_path)

# ==============================
# SPLIT DATASET  (70 / 15 / 15)
# ==============================
output_path = os.path.join(BASE_DIR, "splits")
classes = ["Parasitized", "Uninfected"]

# Clean old splits to avoid stale data
if os.path.exists(output_path):
    shutil.rmtree(output_path)

for cls in classes:
    files = os.listdir(os.path.join(BASE_DIR, cls))
    # Filter only valid image extensions
    files = [f for f in files if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff'))]
    random.shuffle(files)

    n           = len(files)
    train_end   = int(0.70 * n)
    val_end     = int(0.85 * n)

    splits_map = {
        "train": files[:train_end],
        "val"  : files[train_end:val_end],
        "test" : files[val_end:]
    }

    for split_name, split_files in splits_map.items():
        split_dir = os.path.join(output_path, split_name, cls)
        os.makedirs(split_dir, exist_ok=True)
        for file in split_files:
            shutil.copy(
                os.path.join(BASE_DIR, cls, file),
                os.path.join(split_dir, file)
            )

    print(f"{cls}: train={len(splits_map['train'])} | val={len(splits_map['val'])} | test={len(splits_map['test'])}")

print("\nDataset split complete.")

# ==============================
# DATA GENERATORS
# ==============================
train_dir = os.path.join(BASE_DIR, "splits/train")
val_dir   = os.path.join(BASE_DIR, "splits/val")
test_dir  = os.path.join(BASE_DIR, "splits/test")

IMG_SIZE   = (128, 128)
BATCH_SIZE = 32

# Training: augmentation applied
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.15,
    horizontal_flip=True,
    vertical_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

# Validation / Test: only rescale
val_datagen  = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_data = train_datagen.flow_from_directory(
    train_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='binary', seed=SEED
)

val_data = val_datagen.flow_from_directory(
    val_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='binary', seed=SEED
)

test_data = test_datagen.flow_from_directory(
    test_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='binary', shuffle=False
)

print("\nClass Indices:", train_data.class_indices)
# NOTE: class index 0 = Parasitized, 1 = Uninfected  (alphabetical ordering by Keras)

# ==============================
# MODEL 1 — CUSTOM CNN (Baseline)
# ==============================
def build_custom_cnn(input_shape=(128, 128, 3)):
    model = models.Sequential([
        layers.Input(shape=input_shape),

        layers.Conv2D(32, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.25),

        layers.Conv2D(64, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.25),

        layers.Conv2D(128, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.25),

        layers.Conv2D(256, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(2, 2),
        layers.Dropout(0.25),

        layers.Flatten(),
        layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(1e-4)),
        layers.Dropout(0.5),
        layers.Dense(1, activation='sigmoid')
    ], name="CustomCNN")
    return model

# ==============================
# MODEL 2 — TRANSFER LEARNING (MobileNetV2)
# ==============================
def build_mobilenetv2(input_shape=(128, 128, 3)):
    base_model = MobileNetV2(
        input_shape=input_shape,
        include_top=False,
        weights='imagenet'
    )
    # Freeze the base layers initially
    base_model.trainable = False

    inputs  = layers.Input(shape=input_shape)
    x       = base_model(inputs, training=False)
    x       = layers.GlobalAveragePooling2D()(x)
    x       = layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(1e-4))(x)
    x       = layers.Dropout(0.4)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)

    model = models.Model(inputs, outputs, name="MobileNetV2_Transfer")
    return model, base_model

# ==============================
# CALLBACKS
# ==============================
def get_callbacks(model_name):
    return [
        EarlyStopping(
            monitor='val_loss',
            patience=5,
            restore_best_weights=True,
            verbose=1
        ),
        ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.3,
            patience=3,
            min_lr=1e-7,
            verbose=1
        ),
        ModelCheckpoint(
            filepath=f"{model_name}_best.keras",
            monitor='val_accuracy',
            save_best_only=True,
            verbose=0
        )
    ]

# ==============================
# TRAIN CUSTOM CNN
# ==============================
print("\n" + "="*50)
print("TRAINING MODEL 1: Custom CNN")
print("="*50)

cnn_model = build_custom_cnn()
cnn_model.compile(optimizer=Adam(1e-3), loss='binary_crossentropy', metrics=['accuracy'])
cnn_model.summary()

cnn_history = cnn_model.fit(
    train_data,
    validation_data=val_data,
    epochs=20,
    callbacks=get_callbacks("custom_cnn"),
    verbose=1
)

# ==============================
# TRAIN MOBILENETV2 (Feature Extraction → Fine-Tune)
# ==============================
print("\n" + "="*50)
print("TRAINING MODEL 2: MobileNetV2 Transfer Learning")
print("="*50)

mob_model, base_model = build_mobilenetv2()
mob_model.compile(optimizer=Adam(1e-3), loss='binary_crossentropy', metrics=['accuracy'])
mob_model.summary()

# Phase 1: Feature extraction (frozen base)
print("\n[Phase 1] Feature extraction — base frozen")
mob_history_1 = mob_model.fit(
    train_data,
    validation_data=val_data,
    epochs=10,
    callbacks=get_callbacks("mobilenet_phase1"),
    verbose=1
)

# Phase 2: Fine-tuning — unfreeze last 30 layers
print("\n[Phase 2] Fine-tuning — unfreezing last 30 layers")
base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False

mob_model.compile(optimizer=Adam(1e-5), loss='binary_crossentropy', metrics=['accuracy'])

mob_history_2 = mob_model.fit(
    train_data,
    validation_data=val_data,
    epochs=10,
    callbacks=get_callbacks("mobilenet_phase2"),
    verbose=1
)

# Merge both phase histories for plotting
def merge_histories(h1, h2):
    merged = {}
    for key in h1.history:
        merged[key] = h1.history[key] + h2.history[key]
    return merged

mob_history_merged = merge_histories(mob_history_1, mob_history_2)

# ==============================
# PLOT TRAINING CURVES
# ==============================
def plot_training_curves(history_dict, title_prefix, save_name):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle(f"{title_prefix} — Training Curves", fontsize=13, fontweight='bold')

    # Accuracy
    axes[0].plot(history_dict['accuracy'],     label='Train Acc',  color='steelblue',  lw=2)
    axes[0].plot(history_dict['val_accuracy'], label='Val Acc',    color='darkorange', lw=2, linestyle='--')
    axes[0].set_title('Accuracy vs Epoch')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Accuracy')
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # Loss
    axes[1].plot(history_dict['loss'],     label='Train Loss', color='steelblue',  lw=2)
    axes[1].plot(history_dict['val_loss'], label='Val Loss',   color='darkorange', lw=2, linestyle='--')
    axes[1].set_title('Loss vs Epoch')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].legend()
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

plot_training_curves(cnn_history.history,   "Custom CNN",     "cnn_training_curves.png")
plot_training_curves(mob_history_merged,    "MobileNetV2",    "mobilenet_training_curves.png")

# ==============================
# EVALUATION HELPER
# ==============================
def get_predictions(model, data_generator):
    data_generator.reset()
    y_prob = model.predict(data_generator, verbose=0).flatten()
    y_pred = (y_prob > 0.5).astype(int)
    y_true = data_generator.classes
    return y_true, y_pred, y_prob

# ==============================
# CONFUSION MATRIX (ENHANCED)
# ==============================
def plot_confusion_matrix(y_true, y_pred, model_name, save_name):
    cm     = confusion_matrix(y_true, y_pred)
    cm_pct = cm.astype(float) / cm.sum(axis=1, keepdims=True) * 100

    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f"Confusion Matrix — {model_name}", fontsize=13, fontweight='bold')
    labels = ['Parasitized', 'Uninfected']

    # Raw counts
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=labels, yticklabels=labels, ax=axes[0], linewidths=0.5)
    axes[0].set_title('Raw Counts')
    axes[0].set_xlabel('Predicted')
    axes[0].set_ylabel('Actual')

    # Percentage
    sns.heatmap(cm_pct, annot=True, fmt='.1f', cmap='Oranges',
                xticklabels=labels, yticklabels=labels, ax=axes[1], linewidths=0.5)
    axes[1].set_title('Percentage (%)')
    axes[1].set_xlabel('Predicted')
    axes[1].set_ylabel('Actual')

    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")
    return cm

# ==============================
# ROC CURVE
# ==============================
def plot_roc_curve(y_true_list, y_prob_list, model_names, save_name):
    plt.figure(figsize=(8, 6))
    colors = ['steelblue', 'darkorange', 'green', 'red']

    for y_true, y_prob, name, color in zip(y_true_list, y_prob_list, model_names, colors):
        fpr, tpr, _ = roc_curve(y_true, y_prob)
        roc_auc     = auc(fpr, tpr)
        plt.plot(fpr, tpr, color=color, lw=2, label=f"{name} (AUC = {roc_auc:.4f})")

    plt.plot([0, 1], [0, 1], 'k--', lw=1.5, label='Random Classifier')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.02])
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate (Recall)', fontsize=12)
    plt.title('ROC Curves — Model Comparison', fontsize=13, fontweight='bold')
    plt.legend(loc='lower right')
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

# ==============================
# PRECISION-RECALL CURVE
# ==============================
def plot_precision_recall_curve(y_true_list, y_prob_list, model_names, save_name):
    plt.figure(figsize=(8, 6))
    colors = ['steelblue', 'darkorange', 'green', 'red']

    for y_true, y_prob, name, color in zip(y_true_list, y_prob_list, model_names, colors):
        precision, recall, _ = precision_recall_curve(y_true, y_prob)
        ap = average_precision_score(y_true, y_prob)
        plt.plot(recall, precision, color=color, lw=2, label=f"{name} (AP = {ap:.4f})")

    plt.xlabel('Recall', fontsize=12)
    plt.ylabel('Precision', fontsize=12)
    plt.title('Precision-Recall Curves — Model Comparison', fontsize=13, fontweight='bold')
    plt.legend(loc='upper right')
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

# ==============================
# CLASSIFICATION REPORT
# ==============================
def print_classification_report(y_true, y_pred, model_name):
    print(f"\n{'='*50}")
    print(f"Classification Report — {model_name}")
    print(f"{'='*50}")
    print(classification_report(
        y_true, y_pred,
        target_names=['Parasitized', 'Uninfected'],
        digits=4
    ))

# ==============================
# METRICS BAR CHART COMPARISON
# ==============================
def plot_metrics_comparison(results_dict, save_name):
    """
    results_dict: {model_name: {metric_name: value}}
    """
    metrics = ['accuracy', 'precision', 'recall', 'f1_score', 'auc']
    model_names = list(results_dict.keys())
    x = np.arange(len(metrics))
    width = 0.35

    fig, ax = plt.subplots(figsize=(12, 6))
    colors = ['steelblue', 'darkorange']

    for i, (model_name, scores) in enumerate(results_dict.items()):
        vals = [scores.get(m, 0) for m in metrics]
        bars = ax.bar(x + i * width, vals, width, label=model_name, color=colors[i], alpha=0.85)
        for bar, val in zip(bars, vals):
            ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.005,
                    f'{val:.3f}', ha='center', va='bottom', fontsize=9)

    ax.set_xlabel('Metric', fontsize=12)
    ax.set_ylabel('Score', fontsize=12)
    ax.set_title('Model Comparison — Performance Metrics', fontsize=13, fontweight='bold')
    ax.set_xticks(x + width / 2)
    ax.set_xticklabels([m.replace('_', ' ').title() for m in metrics])
    ax.set_ylim(0, 1.12)
    ax.legend()
    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

# ==============================
# RUN FULL EVALUATION — CNN
# ==============================
print("\n" + "="*50)
print("EVALUATING: Custom CNN")
print("="*50)
cnn_true, cnn_pred, cnn_prob = get_predictions(cnn_model, test_data)

cnn_cm = plot_confusion_matrix(cnn_true, cnn_pred, "Custom CNN", "cnn_confusion_matrix.png")
print_classification_report(cnn_true, cnn_pred, "Custom CNN")

cnn_loss, cnn_acc = cnn_model.evaluate(test_data, verbose=0)
print(f"\nTest Loss     : {cnn_loss:.4f}")
print(f"Test Accuracy : {cnn_acc:.4f}")

# ==============================
# RUN FULL EVALUATION — MOBILENETV2
# ==============================
print("\n" + "="*50)
print("EVALUATING: MobileNetV2")
print("="*50)
mob_true, mob_pred, mob_prob = get_predictions(mob_model, test_data)

mob_cm = plot_confusion_matrix(mob_true, mob_pred, "MobileNetV2", "mobilenet_confusion_matrix.png")
print_classification_report(mob_true, mob_pred, "MobileNetV2")

mob_loss, mob_acc = mob_model.evaluate(test_data, verbose=0)
print(f"\nTest Loss     : {mob_loss:.4f}")
print(f"Test Accuracy : {mob_acc:.4f}")

# ==============================
# ROC & PRECISION-RECALL CURVES
# ==============================
plot_roc_curve(
    [cnn_true, mob_true],
    [cnn_prob, mob_prob],
    ["Custom CNN", "MobileNetV2"],
    "roc_curves.png"
)

plot_precision_recall_curve(
    [cnn_true, mob_true],
    [cnn_prob, mob_prob],
    ["Custom CNN", "MobileNetV2"],
    "precision_recall_curves.png"
)

# ==============================
# METRICS COMPARISON BAR CHART
# ==============================
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

def compute_metrics(y_true, y_pred, y_prob):
    return {
        'accuracy' : accuracy_score(y_true, y_pred),
        'precision': precision_score(y_true, y_pred, zero_division=0),
        'recall'   : recall_score(y_true, y_pred, zero_division=0),
        'f1_score' : f1_score(y_true, y_pred, zero_division=0),
        'auc'      : roc_auc_score(y_true, y_prob)
    }

results = {
    "Custom CNN"  : compute_metrics(cnn_true, cnn_pred, cnn_prob),
    "MobileNetV2" : compute_metrics(mob_true, mob_pred, mob_prob)
}

plot_metrics_comparison(results, "model_comparison.png")

# ==============================
# GRAD-CAM IMPLEMENTATION
# ==============================
# Grad-CAM highlights which regions in the image the model focused on
# for its prediction — crucial for medical AI explainability.

def get_gradcam_heatmap(model, img_array, last_conv_layer_name, pred_index=None):
    """
    Computes Grad-CAM heatmap for a given image and model.
    img_array: preprocessed image array of shape (1, H, W, C)
    """
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[model.get_layer(last_conv_layer_name).output, model.output]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]

    grads       = tape.gradient(class_channel, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_outputs = conv_outputs[0]
    heatmap      = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap      = tf.squeeze(heatmap)
    heatmap      = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy()


def overlay_gradcam(img_path, heatmap, alpha=0.4):
    """Superimpose Grad-CAM heatmap over the original image."""
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (128, 128))

    heatmap_resized = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap_uint8   = np.uint8(255 * heatmap_resized)
    heatmap_colored = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_colored = cv2.cvtColor(heatmap_colored, cv2.COLOR_BGR2RGB)

    superimposed = cv2.addWeighted(img, 1 - alpha, heatmap_colored, alpha, 0)
    return img, superimposed


def find_last_conv_layer(model):
    """Auto-detect the last Conv2D layer name."""
    for layer in reversed(model.layers):
        if isinstance(layer, tf.keras.layers.Conv2D):
            return layer.name
    raise ValueError("No Conv2D layer found in model.")


def plot_gradcam_grid(model, folder, label, n=4, save_name="gradcam.png"):
    """Plot Grad-CAM for n sample images from a folder."""
    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith(('.png', '.jpg', '.jpeg'))
    ]
    sample_files = random.sample(files, min(n, len(files)))

    try:
        last_conv = find_last_conv_layer(model)
    except ValueError:
        # For MobileNetV2 transfer model, use the last conv inside base
        last_conv = "Conv_1"

    fig, axes = plt.subplots(n, 3, figsize=(12, n * 3.5))
    fig.suptitle(f"Grad-CAM — {label} Samples", fontsize=13, fontweight='bold')

    col_titles = ["Original", "Grad-CAM Overlay", "Heatmap Only"]
    for ax, ct in zip(axes[0], col_titles):
        ax.set_title(ct, fontsize=11, fontweight='bold')

    for i, fname in enumerate(sample_files):
        img_path = os.path.join(folder, fname)

        img_pil   = keras_image.load_img(img_path, target_size=(128, 128))
        img_array = keras_image.img_to_array(img_pil) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        try:
            heatmap = get_gradcam_heatmap(model, img_array, last_conv)
            original, overlaid = overlay_gradcam(img_path, heatmap)

            axes[i, 0].imshow(original)
            axes[i, 1].imshow(overlaid)
            axes[i, 2].imshow(heatmap, cmap='jet')

            for ax in axes[i]:
                ax.axis('off')
        except Exception as e:
            print(f"  Grad-CAM failed for {fname}: {e}")

    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")


print("\nGenerating Grad-CAM visualizations...")
# CNN Grad-CAM on Parasitized samples
plot_gradcam_grid(
    cnn_model,
    os.path.join(BASE_DIR, "splits/test/Parasitized"),
    "CNN — Parasitized",
    n=4,
    save_name="gradcam_cnn_parasitized.png"
)

# CNN Grad-CAM on Uninfected samples
plot_gradcam_grid(
    cnn_model,
    os.path.join(BASE_DIR, "splits/test/Uninfected"),
    "CNN — Uninfected",
    n=4,
    save_name="gradcam_cnn_uninfected.png"
)

# ==============================
# PROBABILITY DISTRIBUTION PLOT
# ==============================
def plot_probability_distribution(y_true, y_prob, model_name, save_name):
    """Shows how confident the model is for each class."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle(f"Prediction Probability Distribution — {model_name}",
                 fontsize=13, fontweight='bold')

    # Left: KDE plot split by class
    for label, color, name in zip([0, 1], ['red', 'green'], ['Parasitized', 'Uninfected']):
        probs_class = y_prob[y_true == label]
        axes[0].hist(probs_class, bins=40, alpha=0.6, color=color,
                     label=f"{name} (n={len(probs_class)})", density=True)
    axes[0].axvline(x=0.5, color='black', linestyle='--', lw=1.5, label='Threshold (0.5)')
    axes[0].set_xlabel('Predicted Probability (Uninfected)  [sigmoid output; class 1 = Uninfected]')
    axes[0].set_ylabel('Density')
    axes[0].set_title('Probability Distribution by True Class')
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # Right: Box plot
    import pandas as pd
    df = pd.DataFrame({'probability': y_prob, 'true_label': y_true})
    df['class'] = df['true_label'].map({0: 'Parasitized', 1: 'Uninfected'})
    df.boxplot(column='probability', by='class', ax=axes[1],
               boxprops=dict(color='steelblue'),
               medianprops=dict(color='red', lw=2))
    axes[1].set_title('Probability Spread by Class')
    axes[1].set_xlabel('True Class')
    axes[1].set_ylabel('Predicted Probability')
    plt.suptitle('')
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

plot_probability_distribution(cnn_true, cnn_prob, "Custom CNN",  "cnn_prob_distribution.png")
plot_probability_distribution(mob_true, mob_prob, "MobileNetV2", "mob_prob_distribution.png")

# ==============================
# SAMPLE PREDICTIONS GRID
# ==============================
def plot_sample_predictions(model, test_dir, class_indices, n_per_class=5, save_name="predictions_grid.png"):
    """Show sample predictions with correct/wrong labels highlighted."""
    inv_class = {v: k for k, v in class_indices.items()}

    fig, axes = plt.subplots(2, n_per_class, figsize=(n_per_class * 2.5, 6))
    fig.suptitle("Sample Predictions (Green=Correct | Red=Wrong)", fontsize=12, fontweight='bold')

    for row_idx, cls_name in enumerate(['Parasitized', 'Uninfected']):
        cls_dir = os.path.join(test_dir, cls_name)
        files   = [f for f in os.listdir(cls_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        samples = random.sample(files, min(n_per_class, len(files)))

        for col_idx, fname in enumerate(samples):
            img_path  = os.path.join(cls_dir, fname)
            img_pil   = keras_image.load_img(img_path, target_size=(128, 128))
            img_array = keras_image.img_to_array(img_pil) / 255.0
            img_input = np.expand_dims(img_array, axis=0)

            prob     = model.predict(img_input, verbose=0)[0][0]
            pred_idx = int(prob > 0.5)
            pred_cls = inv_class[pred_idx]
            correct  = (pred_cls == cls_name)

            ax = axes[row_idx, col_idx]
            ax.imshow(img_pil)
            ax.set_title(
                f"P: {pred_cls}\n({prob:.2f})",
                color='green' if correct else 'red',
                fontsize=8
            )
            ax.axis('off')

        axes[row_idx, 0].set_ylabel(f"True: {cls_name}", fontsize=9, rotation=0,
                                     labelpad=60, va='center')

    plt.tight_layout()
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

plot_sample_predictions(
    mob_model,
    os.path.join(BASE_DIR, "splits/test"),
    train_data.class_indices,
    n_per_class=5,
    save_name="sample_predictions_grid.png"
)

# ==============================
# THRESHOLD OPTIMIZATION
# ==============================
def find_optimal_threshold(y_true, y_prob, model_name):
    """
    Find the threshold that maximizes F1-score.
    Default 0.5 is not always optimal — especially relevant in medical settings
    where you may prefer higher recall (catch all sick patients).
    """
    thresholds = np.arange(0.1, 0.91, 0.01)
    f1_scores  = []
    recalls    = []
    precisions = []

    for thresh in thresholds:
        preds = (y_prob > thresh).astype(int)
        f1_scores.append(f1_score(y_true, preds, zero_division=0))
        recalls.append(recall_score(y_true, preds, zero_division=0))
        precisions.append(precision_score(y_true, preds, zero_division=0))

    best_idx   = np.argmax(f1_scores)
    best_thresh = thresholds[best_idx]
    best_f1     = f1_scores[best_idx]

    print(f"\n[{model_name}] Optimal Threshold: {best_thresh:.2f} | Best F1: {best_f1:.4f}")

    fig, ax = plt.subplots(figsize=(9, 5))
    ax.plot(thresholds, f1_scores,  label='F1 Score',  color='steelblue',  lw=2)
    ax.plot(thresholds, recalls,    label='Recall',     color='darkorange', lw=2, linestyle='--')
    ax.plot(thresholds, precisions, label='Precision',  color='green',      lw=2, linestyle=':')
    ax.axvline(x=best_thresh, color='red', linestyle='--', lw=1.5,
               label=f'Optimal Threshold ({best_thresh:.2f})')
    ax.set_xlabel('Decision Threshold')
    ax.set_ylabel('Score')
    ax.set_title(f'Threshold Optimization — {model_name}', fontsize=13, fontweight='bold')
    ax.legend()
    ax.grid(alpha=0.3)
    plt.tight_layout()
    save_name = f"threshold_optimization_{model_name.lower().replace(' ', '_')}.png"
    plt.savefig(save_name, dpi=150)
    plt.show()
    print(f"Saved: {save_name}")

    return best_thresh

best_thresh_cnn = find_optimal_threshold(cnn_true, cnn_prob, "Custom CNN")
best_thresh_mob = find_optimal_threshold(mob_true, mob_prob, "MobileNetV2")

# Re-evaluate MobileNetV2 at its optimal threshold
print(f"\n[MobileNetV2] Re-evaluated at optimal threshold ({best_thresh_mob:.2f}):")
mob_pred_optimal = (mob_prob > best_thresh_mob).astype(int)
print(classification_report(
    mob_true, mob_pred_optimal,
    target_names=['Parasitized', 'Uninfected'],
    digits=4
))

# ==============================
# FINAL SUMMARY TABLE
# ==============================
print("\n" + "="*60)
print("FINAL MODEL COMPARISON SUMMARY")
print("="*60)

headers = ["Model", "Accuracy", "Precision", "Recall", "F1", "AUC"]
row_fmt = "{:<20} {:<10} {:<12} {:<10} {:<10} {:<10}"
print(row_fmt.format(*headers))
print("-" * 60)

for model_name, scores in results.items():
    print(row_fmt.format(
        model_name,
        f"{scores['accuracy']:.4f}",
        f"{scores['precision']:.4f}",
        f"{scores['recall']:.4f}",
        f"{scores['f1_score']:.4f}",
        f"{scores['auc']:.4f}"
    ))

print("="*60)

# ==============================
# SAVE MODELS
# ==============================
cnn_model.save("malaria_cnn_model.keras")
mob_model.save("malaria_mobilenet_model.keras")
print("\nModels saved: malaria_cnn_model.keras | malaria_mobilenet_model.keras")

# ==============================
# SINGLE IMAGE INFERENCE FUNCTION
# ==============================
def predict_single_image(model_to_use, img_path, threshold=0.5):
    """
    Run inference on a single image.
    Prints class, confidence, and shows the image.
    """
    img_pil   = keras_image.load_img(img_path, target_size=(128, 128))
    img_array = keras_image.img_to_array(img_pil) / 255.0
    img_input = np.expand_dims(img_array, axis=0)

    # Sigmoid output = P(Uninfected), because Keras indexes classes alphabetically
    # (0 = Parasitized, 1 = Uninfected).
    prob       = model_to_use.predict(img_input, verbose=0)[0][0]
    pred_class = "Uninfected" if prob > threshold else "Parasitized"
    confidence = prob if prob > threshold else 1 - prob

    print(f"\nPrediction : {pred_class}")
    print(f"Confidence : {confidence * 100:.2f}%")
    print(f"Raw Prob   : {prob:.4f}")

    plt.figure(figsize=(4, 4))
    plt.imshow(img_pil)
    plt.title(f"Prediction: {pred_class} ({confidence*100:.1f}%)",
              color='red' if pred_class == 'Parasitized' else 'green',
              fontsize=12, fontweight='bold')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    return pred_class, confidence

# --- Example usage ---
test_sample_dir = os.path.join(BASE_DIR, "splits/test/Parasitized")
sample_image    = os.path.join(test_sample_dir, os.listdir(test_sample_dir)[0])
predict_single_image(mob_model, sample_image, threshold=best_thresh_mob)

print("\n✓ All outputs saved. Project complete.")
```
